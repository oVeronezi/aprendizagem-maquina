### Programação Orientada a Objetos (POO) em python

### 1. Classes e Objetos 
A Classe é o molde (blueprint), enquanto o Objeto é a instancia real de molde.

```python
class Pessoa:
    # O metodo __init__ é p contrutor
    def __init__(self, _nome, _idade):
        self.nome = _nome
        self.idade = _idade

    def saudar(self):
        return f"Olá, meu nome é {self.nome} e tenho {self.idade} anos."

# instanciando objetos
p1 = Pessoa("Vero", 20)
print(p1.saudar())
```

### 2. Encapsulamento
Uso de atributos privados (prefixados com __) para proteger os dados

```python
class ContaBancaria:
    def __init__(self, saldo_inicial):
        self.__saldo = saldo_inicial # Atributo privado

    def depositar(self, valor):
        if valor > 0:
            self.__saldo += valor

    def consultar_saldo(self):  
        return f"Saldo atual: R$ {self.__saldo:.2f}"

conta = ContaBancaria(1000)
conta.depositar(500)
print(conta.consultar_saldo())              
```

### 3. Herança
Permite que uma classe filha herde atributos e métodos de uma classe pai.

```python
class Funcionario(Pessoa):
    def __init__(self, _nome, _idade, _cargo):
        super().__init__(_nome, _idade) # Chama o construtor da classe pai
        self.cargo = _cargo

    def saudar(self):
        return f"Olá sou {self.nome}, atuo como {self.cargo}."

professor = Funcionario("Tiago", 35, "Professor")        
print(professor.saudar())
```

### 4. Polimorfismo
Capacidade de diferentes classes tratarem o mesmo método de formas distintas.

```python
class Animal:
    def falar(self):
        pass # ou ...

class Cachorro(Animal):
    def falar(self):
        return "AU AU!"

class Gato(Animal):
    def falar(self):
        return "MIAUUU!"      

animais = [Cachorro(), Gato()]
for animal in animais:
    print(animal.falar())
```
