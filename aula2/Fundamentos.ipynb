## Fundamentos de Python para Aprendizagem de Maquina

### Estrutura de dados (listas, tuplas, dicionarios)

essas as estruturas mais utilizadas para organizar informações

```python
# Lista (Mutável)
frutas = ["maça", "banana", "cereja",]
frutas.append("uva")

# Tupla (imutável)
coordenadas = (10, "Vini")

# Dicionario (Chave-Valor)
aluno = {"nome": "Tiago", "curso": "Tecnologia", "nota": 9.5}

print(f"Lista: {frutas}")
print(f"Dicionario: {aluno['nome']} tirou {aluno['nota']}")
```

### Estrutura de controle e Repetição

Controle do fluxo de execução do programa

```python
# Condicional
media = 7.5
if media >= 6:
    resultado = "Aprovado"
else:
    resultado = "Exame"

# Loop For
numeros = [1, 2, 3, 4, 5]
quadrados = [x**2 for x in numeros]

print(f"Resultado {resultado}")
print(f"Quadrados {quadrados}")

print("/n Lista de Numeros (jeito longo)") 
quadrados = []
for x in numeros:
    quadrados.append(x**2)
    print(quadrados)    
```

### Funções e Lambda

Modularização de codigo e funções anonimas

```python
def calcular_bonus(salario, percentual=10):
    return salario + (salario * percentual / 100)

# Função Lambda (Anonima)
dobro = lambda  x: x * 2

print(f"Bonus: {calcular_bonus(3000)}")
print(f"Dobro de 5: {dobro(5)}")
```
