```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# A partir de uma lista
meu_array = np.array([1, 2, 3, 4, 5])
print("Meu array:", meu_array)

# Criando um array de zeros
zeros = np.zeros((2, 3)) # Matriz 2x3 de zeros
print("Array de zeros: \n", zeros)

#Criando um array com um intervalo
intervalo = np.arange(0, 11, 2) # de 0 a 10, de 2 em 2
print("Array com intervalo: \n", intervalo)
```

```python
array_a = np.array([1, 2, 3])
array_b = np.array([4, 5, 6])

soma = array_a + array_b
multiplicacao = array_a * 2
multiplicacao_array = array_a * array_b

print("Soma:", soma)
print("Multiplicação:", multiplicacao)
print("Multiplicação de array:", multiplicacao_array)
```

```python
anos = [2020, 2021, 2022, 2023, 2024]
vendas = [100, 120, 150, 130, 160]

plt.plot(anos, vendas)
plt.title("Vendas ao longo dos anos")
plt.xlabel("Ano")
plt.ylabel("Vendas em (milhões)")
plt.grid(True)
plt.show
```

```python
produtos = ['Produto A', 'Produto B', 'Produto C']
quantidades = [50, 80, 65]

plt.bar(produtos, quantidades)
plt.title('Quantidade de Produtos em Estoque')
plt.xlabel('Produto')
plt.ylabel('Quantidade')
plt.show
```

```python
np.random.seed(42)
x = np.random.rand(50) * 10
y = 2 * x + np.random.randn(50) * 2

plt.scatter(x, y)
plt.title('Relação entre X e Y')
plt.xlabel('Variavel X')
plt.ylabel('Variavel Y')
plt.show()
```

```python
# Gerando dados com Numpy
x = np.linspace(0, 2 * np.pi, 100)
y_seno = np.sin(x)
y_cosseno = np.cos(x)

# Criando o gráfico com Matplotlib
plt.plot(x, y_seno, label='Seno')
plt.plot(x, y_cosseno, label='Cosseno')
plt.title('Funções Seno e Cosseno')
plt.xlabel('Angulo [rad]')
plt.ylabel('Valor')
plt.legend()
plt.grid(True)
plt.show
```

```python
# gerando 1000 numeros aleatorios com média (loc) 0 e desvio padrão (scale) 1
dados = np.random.normal(loc=0, scale=1, size=1000)

# cria o histgorama
# 'bins' define o numero de barras no grafico
plt.hist(dados, bins=30, color='skyblue', edgecolor='black')

# adiciona titulos e rotulos
plt.title('Histograma de uma Distribuição Normal')
plt.xlabel('Valores')
plt.ylabel('Frequencia')

# exibe o grafico
plt.show()
```

```python
# cria um array de 100 pontos de -2*PI a 2*PI
x = np.linspace(-2 * np.pi, 2 * np.pi, 200)

# cria uma figura e uam grade de subplot 2x2
# 'fig' é a figura inteira 'axs' é um array 2x2 com os eixos de cada grafico
fig, axs = plt.subplots(2, 2, figsize=(10,8))

#adiciona um titulo principal para todos os subplots
fig.suptitle('Multiplos Gráficos (Subplot)', fontsize= 16)

# grafico 1 (canto superios esquerdo)
axs[0, 0].plot(x, np.sin(x), color='blue')
axs[0, 0].set_title('seno')
axs[0, 0].grid(True)

# grafico 2 (canto superios direito)
axs[0, 1].plot(x, np.cos(x), color='red')
axs[0, 1].set_title('Cosseno')
axs[0, 1].grid(True)

# grafico 3 (canto inferior esquerdo)
axs[1, 0].plot(x, np.tan(x), color='green')
axs[1, 0].set_title('Tangente')
axs[1, 0].set_ylim(-5, 5) # limita o eixo y para melhorar visualização
axs[1, 0].grid(True)

#grafico 4 canto inferior direito
axs[1, 1].plot(x, x**2, color='purple')
axs[1, 1].set_title('Parabola (x^2)')
axs[1, 1].grid(True)

# ajusta o layout para evitar sobreposição de titulos
plt.tight_layout(rect=[0, 0, 1, 0.96])

plt.show()
```

```python
# dados categoricos
categorias = ['Tecnologia', 'Saúde', 'Educação', 'Transporte']
valores = [45, 25, 15, 15] # os valores devem somar 100 para um percentual completo

# 'explode' destaca uma fatia do gráfico. O valor indica o quão distante do centro.
explode = (0.1, 0, 0, 0) # destaca apenas a primeira fatia ('Tecnologia')

# cria o grafico de pizza
plt.pie(valores, explode=explode, labels=categorias, autopct='%1.1f%%',
         shadow=True, startangle=90)

# 'autopct' formata os numeros para mostrar como percentagens
# 'shadow' adiciona uma sombra
# 'startangle' define o angulo inicial

# garante que o grafico seja um circulo
plt.axis('equal')

plt.title('distribuicao de investimentos por setor')
plt.show()
```
