## Exercício 1 — Nota final a partir de horas de estudo e faltas

Uma escola quer prever a nota final de um aluno (0 a 10) a partir das horas de estudo semanais
($x_1$) e do número de faltas no mês ($x_2$):

| Aluno | Horas de estudo ($x_1$) | Faltas ($x_2$) | Nota final (y) |
|---|---|---|---|
| A | 10 | 1 | 7 |
| B | 6 | 3 | 3 |
| C | 8 | 0 | 7 |

**Perguntas:** a) monte X e y; b) calcule $\beta$; c) nota prevista para 12h de estudo e 2 faltas;
d) interprete os sinais de $\beta_1$ e $\beta_2$; e) por que Múltipla e não Simples?

```python
# 1. Montando a matriz de features e o vetor alvo
X_features = np.array([[10, 1], [6, 3], [8, 0]])  # colunas: horas de estudo, faltas
y = np.array([7, 3, 7])

# 2. Adicionando o termo de intercepto (bias): coluna de 1's à esquerda
m = X_features.shape[0]  # número de exemplos
X = np.c_[np.ones(m), X_features]

# 3. Aplicando a equação normal: beta = (X.T @ X)^-1 @ X.T @ y
beta = np.linalg.inv(X.T @ X) @ X.T @ y

print("=== Coeficientes Descobertos ===")
print(f"Intercepto (Beta 0): {beta[0]:.2f}")
print(f"Peso para Horas de Estudo (Beta 1): {beta[1]:.2f}")
print(f"Peso para Faltas (Beta 2): {beta[2]:.2f}")
```

```python
# 4. Fazendo uma previsão para um aluno com 12h de estudo e 2 faltas
novo_aluno = np.array([1, 12, 2])  # o '1' inicial representa o intercepto
nota_prevista = novo_aluno @ beta

print("=== Teste de Previsão ===")
print(f"Novo aluno: 12h de estudo, 2 faltas")
print(f"Nota prevista pelo modelo: {nota_prevista:.2f}")

# 5. Avaliando o ajuste do modelo com R²
y_pred = X @ beta
sqr = np.sum((y - y_pred) ** 2)
sqt = np.sum((y - np.mean(y)) ** 2)
r2 = 1 - (sqr / sqt)
print(f"\nCoeficiente de Determinação (R²): {r2:.4f}")
```

### Respostas — Exercício 1

**a) e b)** $X$ e $y$ montados e $\beta$ calculado na célula acima: $\beta_0=3$, $\beta_1=0{,}5$,
$\beta_2=-1$ → **modelo: nota = 3 + 0,5 × horas − 1 × faltas**.

**c)** Nota prevista = **7,00** (calculado na célula acima).

**d)** $\beta_1=0{,}5>0$: mais horas de estudo aumentam a nota (esperado). $\beta_2=-1<0$: cada
falta reduz a nota prevista em 1 ponto (esperado).

**e)** Porque há **duas** variáveis explicativas (horas e faltas) influenciando a nota ao mesmo
tempo — Regressão Simples só suporta uma variável.

## Exercício 2 — Aluguel a partir de área e distância ao centro

Uma imobiliária quer prever o aluguel (R$) a partir da área (m², $x_1$) e da distância ao centro
(km, $x_2$):

| Imóvel | Área ($x_1$) | Distância ($x_2$) | Aluguel (y) |
|---|---|---|---|
| A | 40 | 5 | 750 |
| B | 60 | 2 | 1080 |
| C | 30 | 10 | 550 |

**Perguntas:** a) calcule $\beta_0,\beta_1,\beta_2$; b) escreva o modelo; c) aluguel previsto para
imóvel de 50 m² a 4 km do centro; d) o sinal de $\beta_2$ faz sentido?

```python
# 1. Montando a matriz de features e o vetor alvo
X_features = np.array([[40, 5], [60, 2], [30, 10]])  # colunas: área, distância ao centro
y = np.array([750, 1080, 550])

# 2. Adicionando o termo de intercepto (bias)
m = X_features.shape[0]
X = np.c_[np.ones(m), X_features]

# 3. Aplicando a equação normal
beta = np.linalg.inv(X.T @ X) @ X.T @ y

print("=== Coeficientes Descobertos ===")
print(f"Intercepto (Beta 0): {beta[0]:.2f}")
print(f"Peso para Área (Beta 1): {beta[1]:.2f}")
print(f"Peso para Distância (Beta 2): {beta[2]:.2f}")
```

```python
# 4. Fazendo uma previsão para um imóvel de 50 m² a 4 km do centro
novo_imovel = np.array([1, 50, 4])
aluguel_previsto = novo_imovel @ beta

print("=== Teste de Previsão ===")
print(f"Novo imóvel: 50 m², 4 km do centro")
print(f"Aluguel previsto pelo modelo: R$ {aluguel_previsto:.2f}")

# 5. Avaliando o ajuste do modelo com R²
y_pred = X @ beta
sqr = np.sum((y - y_pred) ** 2)
sqt = np.sum((y - np.mean(y)) ** 2)
r2 = 1 - (sqr / sqt)
print(f"\nCoeficiente de Determinação (R²): {r2:.4f}")
```

### Respostas — Exercício 2

**a) e b)** $\beta$ calculado na célula acima: $\beta_0=200$, $\beta_1=15$, $\beta_2=-10$ →
**modelo: aluguel = 200 + 15 × área − 10 × distância**.

**c)** Aluguel previsto = **R\$ 910,00** (calculado na célula acima).

**d)** Sim — $\beta_2=-10<0$ significa que, quanto mais longe do centro, menor o aluguel
(mantendo a área constante), o que é o esperado no mercado imobiliário.
