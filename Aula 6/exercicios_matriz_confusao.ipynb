# Exercícios de Matriz de Confusão — Resolução

Resolução dos exercícios do **Guia de Estudos - Prova.md**, seguindo o mesmo padrão de código
usado na Aula 6 (métricas de avaliação calculadas com Numpy puro, a partir de VP/VN/FP/FN).

```python
# =======================
# funções de avaliação
# =======================

def calculate_metrics(VP, VN, FP, FN):
    acc = (VP + VN) / (VP + VN + FP + FN)
    prec = VP / (VP + FP)
    rec = VP / (VP + FN)
    spec = VN / (VN + FP)
    f1 = 2 * (prec * rec) / (prec + rec)
    return acc, prec, rec, spec, f1
```

## Exercício 1 — Métricas básicas

Um modelo de classificação de spam gerou a seguinte matriz de confusão: VP=40, VN=50, FP=5, FN=5.

**Pergunta:** calcule acurácia, precisão, recall, especificidade e F1.

```python
VP, VN, FP, FN = 40, 50, 5, 5

acc, prec, rec, spec, f1 = calculate_metrics(VP, VN, FP, FN)

print("=" * 30)
print(" MATRIZ DE CONFUSÃO ")
print("=" * 30)
print(f"VERDADEIROS POSITIVOS (VP): {VP}")
print(f"VERDADEIROS NEGATIVOS (VN): {VN}")
print(f"FALSOS POSITIVOS (FP): {FP}")
print(f"FALSOS NEGATIVOS (FN): {FN}")

print("\n" + "=" * 30)
print(" MÉTRICAS DE AVALIAÇÃO ")
print("=" * 30)
print(f"Acurácia:       {acc:.4f} ({(acc * 100):.1f}%)")
print(f"Precisão:       {prec:.4f} ({(prec * 100):.1f}%)")
print(f"Revocação:      {rec:.4f} ({(rec * 100):.1f}%)")
print(f"Especificidade: {spec:.4f} ({(spec * 100):.1f}%)")
print(f"F1-Score:       {f1:.4f} ({(f1 * 100):.1f}%)")
```

### Respostas — Exercício 1

Acurácia = 0,90 · Precisão = 0,889 · Recall = 0,889 · Especificidade = 0,909 · F1 = 0,889
(valores calculados na célula acima).

## Exercício 2 — A armadilha da acurácia (dados desbalanceados)

Um exame para detectar uma doença rara foi aplicado em 200 pessoas: VP=8, VN=180, FP=2, FN=10.

**Perguntas:** a) calcule todas as métricas; b) a acurácia de ~94% indica que o exame é excelente?
c) nesse contexto, qual métrica importa mais: precisão ou recall?

```python
VP, VN, FP, FN = 8, 180, 2, 10

acc, prec, rec, spec, f1 = calculate_metrics(VP, VN, FP, FN)

print("=" * 30)
print(" MATRIZ DE CONFUSÃO ")
print("=" * 30)
print(f"VERDADEIROS POSITIVOS (VP): {VP}")
print(f"VERDADEIROS NEGATIVOS (VN): {VN}")
print(f"FALSOS POSITIVOS (FP): {FP}")
print(f"FALSOS NEGATIVOS (FN): {FN}")

print("\n" + "=" * 30)
print(" MÉTRICAS DE AVALIAÇÃO ")
print("=" * 30)
print(f"Acurácia:       {acc:.4f} ({(acc * 100):.1f}%)")
print(f"Precisão:       {prec:.4f} ({(prec * 100):.1f}%)")
print(f"Revocação:      {rec:.4f} ({(rec * 100):.1f}%)")
print(f"Especificidade: {spec:.4f} ({(spec * 100):.1f}%)")
print(f"F1-Score:       {f1:.4f} ({(f1 * 100):.1f}%)")

print("\n--> Apesar da acurácia alta (94%), a revocação de apenas 44,4% mostra que o exame")
print("    deixa passar mais da metade dos doentes reais (10 de 18). Em doenças, a Revocação")
print("    (Recall) é a métrica mais importante: um falso negativo é bem mais grave que um falso positivo.")
```

### Respostas — Exercício 2

**a)** Acurácia = 0,94 (94%) · Precisão = 0,80 (80%) · Recall ≈ 0,444 (44,4%) ·
Especificidade ≈ 0,989 (98,9%) · F1 ≈ 0,571 (valores calculados na célula acima).

**b)** **Não.** A acurácia de 94% parece ótima só porque a maioria das pessoas é saudável (a
classe negativa domina o dataset). A recall de apenas 44,4% revela que o exame **deixa passar
mais da metade dos doentes reais** (10 de 18) — um resultado péssimo para um exame médico.

**c)** **Recall** é a métrica mais importante aqui: em doenças, o custo de um falso negativo
(deixar um doente sem diagnóstico) é muito mais grave que o de um falso positivo (mandar uma
pessoa saudável fazer um exame extra de confirmação).
